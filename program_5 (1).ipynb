```python
import pandas as pd

values=[
    48,49,49,50,50,50,51,51,51,52,
    52,52,53,53,53,54,54,55,55,100
]

mean=sum(values)/len(values)
std=pd.Series(values).std(ddof=0)

print("Mean:",round(mean,2))
print("Standard Deviation", round(std,2))

for x in values:
  z=(x-mean)/std
  if abs(z)>3:
    print("Outlier:", x)
    print("Z-score:", round(z,2))
```
